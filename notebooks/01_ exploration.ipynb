```python
import pandas as pd

df = pd.read_csv('../data/raw/zomato.csv', encoding='latin-1')
print(f"Shape: {df.shape}")
print(f"\nColumns:\n{df.columns.tolist()}")
df.head()
```

```python
print(f"Total rows: {df.shape[0]}")
print(f"Total columns: {df.shape[1]}")
print(f"\nData types:")
print(df.dtypes)
```

```python
missing = df.isnull().sum().sort_values(ascending=False)
print("Missing values:")
print(missing[missing > 0])
```

```python
print("Rate column samples:")
print(df['rate'].value_counts().head(10))
```

```python
print(f"Cities: {df['listed_in(city)'].nunique()}")
print(df['listed_in(city)'].value_counts().head(10))

print(f"\nRestaurant types: {df['rest_type'].nunique()}")
print(df['rest_type'].value_counts().head(10))

print(f"\nOnline order:")
print(df['online_order'].value_counts())
```

```python
print("Cost column samples:")
print(df['approx_cost(for two people)'].value_counts().head(10))
```
